```python
import os
import calendar
import time

ERA5_DIR = "/content/era5_data"
os.makedirs(ERA5_DIR, exist_ok=True)

# Make sure the ECMWF client exists
if "client" not in globals():
    raise RuntimeError(
        "ECMWF client is not available. "
        "Please recreate the client before running this cell."
    )

# ERA5 collection period
start_year = 2024
end_year = 2026

# 2026 is available only through June for our project
last_month = {
    2024: 12,
    2025: 12,
    2026: 6
}

variables = [
    "2m_temperature",
    "2m_dewpoint_temperature",
    "surface_pressure"
]

# Bounding box covering all 20 stations
north = float(stations["era5_latitude"].max())
west = float(stations["era5_longitude"].min())
south = float(stations["era5_latitude"].min())
east = float(stations["era5_longitude"].max())

area = [north, west, south, east]

print("ERA5 area:", area)
print("Variables:", variables)
print()

for year in range(start_year, end_year + 1):

    for month in range(1, last_month[year] + 1):

        output_file = (
            f"{ERA5_DIR}/era5_{year}_{month:02d}.nc"
        )

        # Skip files that have already been downloaded
        if os.path.exists(output_file):
            print(
                f"⏭️ Skipping {year}-{month:02d} "
                f"(already exists)"
            )
            continue

        num_days = calendar.monthrange(year, month)[1]

        request = {
            "product_type": "reanalysis",

            "variable": variables,

            "year": str(year),

            "month": f"{month:02d}",

            "day": [
                f"{day:02d}"
                for day in range(1, num_days + 1)
            ],

            "time": [
                f"{hour:02d}:00"
                for hour in range(24)
            ],

            "area": area,

            "data_format": "netcdf",

            "download_format": "unarchived"
        }

        print(
            f"\n⬇️ Downloading ERA5 "
            f"{year}-{month:02d}..."
        )

        try:

            client.retrieve(
                "reanalysis-era5-single-levels",
                request,
                output_file
            )

            print(
                f"✅ Successfully downloaded "
                f"{year}-{month:02d}"
            )

        except Exception as e:

            print(
                f"❌ Failed {year}-{month:02d}"
            )

            print("Error:", e)

            print(
                "\nThe loop will continue with "
                "the next month."
            )

        # Small pause between requests
        time.sleep(2)

print("\n🎉 ERA5 collection loop finished!")
```

```python
import xarray as xr
import numpy as np
import pandas as pd

era5_file = "/content/era5_data/era5_2024_02.nc"

ds = xr.open_dataset(era5_file)

# ERA5 temperature: Kelvin → Celsius
ds["temperature_C"] = ds["t2m"] - 273.15

# ERA5 pressure: Pascal → hPa
ds["pressure_hPa"] = ds["sp"] / 100.0

# Dew point remains in Kelvin for the humidity calculation
# Calculate relative humidity using the Magnus approximation

T = ds["t2m"] - 273.15
Td = ds["d2m"] - 273.15

a = 17.625
b = 243.04

gamma_T = (a * T) / (b + T)
gamma_Td = (a * Td) / (b + Td)

ds["relative_humidity_pct"] = (
    100.0 * np.exp(gamma_Td - gamma_T)
)

print("✅ ERA5 conversion complete")

print("\nVariables:")
print(list(ds.data_vars))

print("\nTemperature:")
print(
    float(ds["temperature_C"].min()),
    "to",
    float(ds["temperature_C"].max()),
    "°C"
)

print("\nPressure:")
print(
    float(ds["pressure_hPa"].min()),
    "to",
    float(ds["pressure_hPa"].max()),
    "hPa"
)

print("\nRelative humidity:")
print(
    float(ds["relative_humidity_pct"].min()),
    "to",
    float(ds["relative_humidity_pct"].max()),
    "%"
)
```

```python
import xarray as xr
import numpy as np
import os

processed_dir = "/content/era5_processed"
os.makedirs(processed_dir, exist_ok=True)

era5_files = sorted(
    f for f in os.listdir("/content/era5_data")
    if f.endswith(".nc")
)

def process_era5_file(input_file, output_file):

    ds = xr.open_dataset(input_file)

    # Kelvin → Celsius
    ds["temperature_C"] = ds["t2m"] - 273.15

    # Pascal → hPa
    ds["pressure_hPa"] = ds["sp"] / 100.0

    # Temperature and dew point in Celsius
    T = ds["t2m"] - 273.15
    Td = ds["d2m"] - 273.15

    # Magnus formula
    a = 17.625
    b = 243.04

    gamma_T = (a * T) / (b + T)
    gamma_Td = (a * Td) / (b + Td)

    rh = 100.0 * np.exp(gamma_Td - gamma_T)

    # Prevent tiny numerical overshoots
    ds["relative_humidity_pct"] = rh.clip(
        min=0,
        max=100
    )

    # Keep only the variables we need
    ds = ds[
        [
            "temperature_C",
            "pressure_hPa",
            "relative_humidity_pct"
        ]
    ]

    ds.to_netcdf(output_file)

    ds.close()


for i, filename in enumerate(era5_files, start=1):

    input_path = os.path.join(
        "/content/era5_data",
        filename
    )

    output_filename = filename.replace(
        ".nc",
        "_processed.nc"
    )

    output_path = os.path.join(
        processed_dir,
        output_filename
    )

    if os.path.exists(output_path):

        print(
            f"⏭️ {i}/{len(era5_files)} "
            f"{filename} already processed"
        )

        continue

    print(
        f"🔄 Processing {i}/{len(era5_files)}: "
        f"{filename}"
    )

    process_era5_file(
        input_path,
        output_path
    )

    print("   ✅ Done")


print("\n🎉 All ERA5 files processed!")
```

```python
import os
import gc
import xarray as xr
import pandas as pd
import numpy as np

ERA5_PROCESSED_DIR = "/content/era5_processed"
HYBRID_DIR = "/content/hybrid_monthly"

os.makedirs(HYBRID_DIR, exist_ok=True)


def process_one_month(year, month, skyguard_data):

    month_str = f"{month:02d}"

    print(f"\n{'='*50}")
    print(f"Processing {year}-{month_str}")
    print(f"{'='*50}")

    # ------------------------------------------
    # 1. Select SkyGuard rows for this month
    # ------------------------------------------

    start = pd.Timestamp(
        f"{year}-{month_str}-01",
        tz="UTC"
    )

    if month == 12:
        end = pd.Timestamp(
            f"{year+1}-01-01",
            tz="UTC"
        )
    else:
        end = pd.Timestamp(
            f"{year}-{month+1:02d}-01",
            tz="UTC"
        )

    sky_month = skyguard_data[
        (skyguard_data["timestamp"] >= start) &
        (skyguard_data["timestamp"] < end)
    ].copy()

    if len(sky_month) == 0:
        print("⚠️ No SkyGuard rows.")
        return None

    print("SkyGuard rows:", len(sky_month))

    # ------------------------------------------
    # 2. Open ERA5
    # ------------------------------------------

    era5_file = os.path.join(
        ERA5_PROCESSED_DIR,
        f"era5_{year}_{month_str}_processed.nc"
    )

    if not os.path.exists(era5_file):
        print("❌ ERA5 file not found:", era5_file)
        return None

    ds = xr.open_dataset(era5_file)

    # ------------------------------------------
    # 3. Convert ERA5 to dataframe
    # ------------------------------------------

    era5_month = ds[
        [
            "temperature_C",
            "pressure_hPa",
            "relative_humidity_pct"
        ]
    ].to_dataframe().reset_index()

    ds.close()

    # ------------------------------------------
    # 4. Match ERA5 grid points to stations
    # ------------------------------------------

    grid_points = stations[
        [
            "station_id",
            "era5_latitude",
            "era5_longitude"
        ]
    ].drop_duplicates()

    era5_month = era5_month.merge(
        grid_points,
        left_on=["latitude", "longitude"],
        right_on=["era5_latitude", "era5_longitude"],
        how="inner"
    )

    # ------------------------------------------
    # 5. Prepare timestamps
    # ------------------------------------------

    sky_month["timestamp"] = pd.to_datetime(
        sky_month["timestamp"],
        utc=True
    ).astype("datetime64[ns, UTC]")

    era5_month["valid_time"] = pd.to_datetime(
        era5_month["valid_time"],
        utc=True
    ).astype("datetime64[ns, UTC]")

    # ------------------------------------------
    # 6. Sort for merge_asof
    # ------------------------------------------

    sky_month = sky_month.sort_values(
        "timestamp"
    ).reset_index(drop=True)

    era5_month = era5_month.sort_values(
        "valid_time"
    ).reset_index(drop=True)

    # ------------------------------------------
    # 7. Temporal matching
    # ------------------------------------------

    matched = pd.merge_asof(
        sky_month,
        era5_month[
            [
                "station_id",
                "valid_time",
                "temperature_C",
                "pressure_hPa",
                "relative_humidity_pct"
            ]
        ],
        left_on="timestamp",
        right_on="valid_time",
        by="station_id",
        direction="nearest",
        tolerance=pd.Timedelta("30min"),
        suffixes=("_skyguard", "_era5")
    )

    # ------------------------------------------
    # 8. Create hybrid differences
    # ------------------------------------------

    matched["temperature_difference"] = (
        matched["temperature_C_skyguard"]
        - matched["temperature_C_era5"]
    )

    matched["pressure_difference"] = (
        matched["pressure_hPa_skyguard"]
        - matched["pressure_hPa_era5"]
    )

    matched["humidity_difference"] = (
        matched["relative_humidity_pct_skyguard"]
        - matched["relative_humidity_pct_era5"]
    )

    # ------------------------------------------
    # 9. Matching quality flag
    # ------------------------------------------

    matched["era5_match"] = (
        matched["valid_time"].notna()
    )

    # ------------------------------------------
    # 10. Save monthly result
    # ------------------------------------------

    output_file = os.path.join(
        HYBRID_DIR,
        f"hybrid_{year}_{month_str}.parquet"
    )

    matched.to_parquet(
        output_file,
        index=False
    )

    print("Hybrid rows:", len(matched))
    print(
        "ERA5 matches:",
        matched["era5_match"].sum()
    )

    print(
        "Saved:",
        output_file
    )

    # Clean memory
    del sky_month
    del era5_month
    del matched

    gc.collect()

    return output_file
```

```python
# ==========================================
# STEP 13 — PROCESS ALL REMAINING MONTHS
# ==========================================

successful = []
failed = []

for year in [2024, 2025, 2026]:

    # 2024 and 2025 → 12 months
    # 2026 → January through June
    max_month = 12 if year < 2026 else 6

    for month in range(1, max_month + 1):

        output_file = (
            f"/content/hybrid_monthly/"
            f"hybrid_{year}_{month:02d}.parquet"
        )

        # January 2024 already completed
        if os.path.exists(output_file):

            print(
                f"⏭️ Skipping {year}-{month:02d} "
                f"(already exists)"
            )

            successful.append(output_file)
            continue

        try:

            result = process_one_month(
                year,
                month,
                hybrid_df
            )

            if result is not None:
                successful.append(result)
            else:
                failed.append(f"{year}-{month:02d}")

        except Exception as e:

            print(
                f"❌ FAILED {year}-{month:02d}"
            )

            print("Error:", e)

            failed.append(
                f"{year}-{month:02d}"
            )

print("\n" + "="*50)
print("🎉 HYBRID PROCESSING FINISHED")
print("="*50)

print("\nSuccessful:", len(successful))
print("Failed:", len(failed))

if failed:
    print("\nFailed months:")
    print(failed)
```
